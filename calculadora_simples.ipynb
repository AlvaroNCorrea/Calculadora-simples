```python
# Soma
def somar (a,b):
  somar = a + b
  return print(f"A soma dos numeros {a} + {b} é igual a {somar}")

# Subtração
def subtrair (a,b):
  subtrair = a - b
  return print(f"A diferença dos numeros {a} - {b} é igual a {subtrair}")

# Multiplicar
def multiplicar (a,b):
  multiplicar = a * b
  return print(f"A multiplicação dos numeros {a} x {b} é igual a {multiplicar}")

# Divisão
def dividir (a,b):
  if b == 0:
    return print("Erro: não é possivel dividir por zero")
  else:
    dividir = a / b
    return print(f"A divisão dos numeros {a} / {b} é igual a {dividir}")

while True:
    print("\n--- Calculadora Python ---")
    print("1 - somar")
    print("2 - subtrair")
    print("3 - multiplicar")
    print("4 - dividir")
    print("0 - sair")

    escolha = (input("Escolha uma opção: "))

    if escolha == "0":
      print("Calculadora encerrada")
      break

    if escolha in ["1", "2", "3", "4"]:
      num1 = float(input("Digite o primeiro número: "))
      num2 = float(input("Digite o segundo número: "))

      if escolha == "1":
        resultado = somar(num1, num2)
      elif escolha == "2":
        resultado = subtrair(num1, num2)
      elif escolha == "3":
        resultado = multiplicar(num1, num2)
      elif escolha == "4":
        resultado = dividir(num1, num2)
    else:
      print("Escolha um numero valido")
```
